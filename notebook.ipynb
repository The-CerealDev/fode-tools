```python
# differential equation
def diff(x: float, y: float) -> float:
    return(x**2+y**2)# dy/dx = x^2 + y^2
```

```python
diff(1,2)
```

```python
def euler(x_n,y_n,dydx_n,step) -> float:
    
    y_next = float(y_n) + float(step) * float(dydx_n)

    return(x_n + step, y_next)
```

```python
euler(1, 1, diff(1, 1), 0.5)
```
